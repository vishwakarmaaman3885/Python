#### Polymorphism-“one name, many forms.”
##### In Python OOP, the same method name can behave differently depending on the object.

```python
def test(a,b):
    return a+b 
```

```python
test(4,5) 
```

```python
test("Aman"," Kumar") 
```

```python
test([1,2,3,4,5],[4,5,6,7]) 
```

```python
class data_science:
    def syllabus(self):
        print("this is my method for data science syllabus") 
```

```python
class web_dev:
    def syllabus(self):
        print('this is my method for web dev') 
```

```python
def class_parcer(class_obj):
    for i in class_obj:
        i.syllabus() 
```

```python
obj_data_science = data_science() 
```

```python
obj_web_dev = web_dev() 
```

```python
class_obj = [obj_data_science,obj_web_dev] 
```

```python
class_parcer(class_obj) 
```
