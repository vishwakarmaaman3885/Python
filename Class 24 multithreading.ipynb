#### Multithreading in Python is a technique that allows multiple threads to make progress concurrently within a single program, helping it perform multiple tasks efficiently.

Example: Downloading a file while playing music.

```python
import threading
```

```python
def test(id) :
    print("this is my test id %d " % id)
```

```python
test(10)
```

```python
test(1)
```

```python
test(3)
```

```python
thread = [threading.Thread(target=test , args = (i,)) for i in [10 , 1 ,3]]
```

```python
thread
```

```python
for t in thread:
    t.start()
```

```python
import urllib.request
def file_download(url , filename):
    urllib.request.urlretrieve(url , filename)
```

```python
import time
```

```python
def test2(x):
    for i in range(10):
        print("tes1 print the value of x %d and print the value of i %d" %(x,i)) 
        time.sleep(1) 
```

```python
thread2 = [threading.Thread(target=test2,args=(i,)) for i in [100,10,20,5]] 
```

```python
for t in thread2:
    t.start() 
```

```python
test2(10) 
```

```python
test2(2)
```

```python
shared_var = 0
lock_var = threading.Lock()
def test4(x):
    global shared_var
    with lock_var:
        shared_var = shared_var+1
        print("value of x%d and value of shared_var  %d " %(x,shared_var))
        time.sleep(1)

thread5 = [threading.Thread(target=test4, args=(i, )) for i in [1,2,3,4,5]]
for t in thread5:
    t.start() 
```

```python
test4(1) 
```

```python
test4(2) 
```
