#### Encapsulation-Hide the data and provide controlled methods to access or modify it.
##### Encapsulation = Data hiding + controlled access.

```python
class test:
    def __init__(self,a,b): # init-Constructor → gives initial values to an object.
        self.a = a
        self.b = b 
```

```python
t = test(45,56) 
```

```python
t.a
```

```python
t.b
```

```python
t.a = 3243
```

```python
t.a 
```

```python
class car:
    def __init__(self,year,make,model,speed):
        self.__year = year # __ = used to restrict direct access to an attribute from outside the class.only admin can access the data
        self.__make = make
        self.__model = model
        self.__speed = 0
    
    def set_speed(self,speed): # This method is used to change the speed.
        self.__speed = 0 if speed<0 else speed # This method is used to read the current speed.
    
    def get_speed(self):
        return self.__speed  
```

```python
obj_car = car(2025,"toyota","innova",160)
```

```python
obj_car.get_speed()  
```

```python
obj_car.set_speed(-1234) 
```

```python
obj_car.set_speed(3456) 
```

```python
obj_car._car__year = 2023
```

```python
obj_car._car__year
```

```python
obj_car._car__speed 
```

```python
class bank_account:
    def __init__(self,balance): # here balance is private
        self.__balance = balance
    def deposit(self,amount):
        self.__balance = self.__balance + amount
    def withdraw(self,amount):
        if self.__balance>=amount:
            self.__balance = self.__balance - amount
            return "Transaction sucessfull"  # hide the entire impelementation of user
        else:
            return "Insufficient balance"
    def get_balance(self):
        return self.__balance 
```

```python
obj_bank_account =  bank_account(1000) 
```

```python
obj_bank_account.get_balance() 
```

```python
obj_bank_account.deposit(4000) 
```

```python
obj_bank_account.get_balance() 
```

```python
obj_bank_account.withdraw(7000) 
```

```python
obj_bank_account.withdraw(2000) 
```

```python
obj_bank_account.get_balance() 
```
