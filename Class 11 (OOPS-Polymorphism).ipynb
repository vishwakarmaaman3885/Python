### Polymorphism

```python
def test(a,b): # The same method name can perform different actions depending on the object.
    return a+b  # Polymorphism = “many forms.”. this is the example of polymorphism
```

```python
test(12,11) 
```

```python
test("Aman"," Kumar") 
```

```python
test([2,3,4,5,5],[4,5,6,7]) 
```

```python
class data_science:
    def syllabus(self):
        print("this is my method for data science syllabus") 
```

```python
class web_dev:
    def syllabus(self):
        print("this is my method for web dev") 
```

```python
def class_parcer(class_obj) : 
    for i in class_obj : 
        i.syllabus() 
```

```python
obj_data_science = data_science() 
```

```python
obj_web_dev = web_dev() 
```

```python
class_obj = [obj_data_science,obj_web_dev] 
```

```python
class_parcer(class_obj) 
```
