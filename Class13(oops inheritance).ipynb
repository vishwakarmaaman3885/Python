#### Inheritance-Child class can reuse the code of Parent class.

```python
# Animal → Dog. Animal = Parent class, Dog = Child class
# Dog is an Animal, so Dog can inherit common behavior from Animal. 
```

```python
class parent:
    def test_parent(self):
        print("this is my parent class") 
```

```python
class child(parent):
    pass
```

```python
child_obj = child() 
```

```python
child_obj.test_parent() 
```

```python
class class1:
    def test_class1(self):
        print("this is my class1") 
```

```python
class class2(class1):
    def test_class2(self):
        print("this is my class2") 
```

```python
class class3(class2): # this is example of multilevel inher. - Grandparent → Parent → Child.
    def test_class3(self):
        print("this is my class3") 
```

```python
obj_class3 = class3() 
```

```python
obj_class3.test_class1()
```

```python
obj_class3.test_class2() 
```

```python
obj_class3.test_class3() 
```

```python
class class1:
    def test_class1(self):
        print("this is my class 1") 
```

```python
class class2:
    def test_class2(self):
        print("this is my class 2") 
```

```python
class class3 (class1,class2): # multilevel inheritance- Multiple parents → one child.
    pass
```

```python
obj_class3 = class3() 
```

```python
obj_class3.test_class1() 
```

```python
obj_class3.test_class2()  
```

```python
# Single Inheritance: One parent class → one child class. A → B
# Multiple Inheritance: Multiple parent classes → one child class.
# Multilevel Inheritance: A class inherits from another derived class.
# Hierarchical Inheritance: One parent class → multiple child classes.
# Hybrid Inheritance: Combination of two or more types of inheritance. 
```
