```python
# Loop- A loop is used to repeat the same code multiple times without writing it again and again.
# While - Used when repetition depends on a condition

joining_age = 25
while joining_age<=60:
    print(joining_age)
    joining_age = joining_age+1
else:
    print("Its time to retirement") 
```

```python
## ATM machine with 1000 rs
Total_amount = 1000
while Total_amount!=0:
    print(Total_amount)
    Total_amount = Total_amount-100
else:
    print("Put more money bank people") 
```

```python
# Nested loops
n = 7

for i in range(0,n):
    for j in range(0,i+1):
        print("*",end = "")
    print("\r")
```

```python
n = 4  # outer loops perform 4 times

for i in range(n):
    for j in range(2 * i + 1):
        print("*", end="")
    print()
```

```python
# Loop controls
# Break - Stop the loop(immidiately)
# Pass- Do nothing
fruits_list = ["Mango","Orange","Apple","Papaya","Cherry"]
```

```python
for i in fruits_list:
    if i == "Apple":
        print("The fruit is apple")
        break
    print(i) 
```

```python
for i in fruits_list:
    if i == "Apple":
        print("The fruit is apple")
        pass
    print(i) 
```

```python
# Continue - Skip the current iteration and move to the next iteration of the loop.
for i in fruits_list:
    if i == "Apple":
        print("The fruit is apple")
        continue
    print(i) 
```
