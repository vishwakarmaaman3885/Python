```python
data = [["name","email_id","number"],
        ["Aman","aman@gmail.com",738935],
        ["Khushi","khushi@gmail.com",99936]
        ]
```

```python
import csv # which is used to read and write CSV (Comma-Separated Values) files.
```

```python
with open("test3.csv","w") as f:
    w = csv.writer(f)
    for i in data:
        w.writerow(i) 
# csv.writer(f)     → creates CSV writer
# writerow(i)       → writes ONE row
# writerows(data)   → writes MANY rows
```

```python
with open("test3.csv","r") as f:
    read = csv.reader(f)
    for i in read:
        print(i) 
```

```python
# Buffered Read And Write & Other File Methods
# Buffering means Python temporarily stores data in memory before actually reading from or writing to the file.
```

```python
import io # Input/Output (I/O) module. It provides tools for working with streams of data, including text and binary data.
```

```python
with open("test.text4","wb") as f:
    file = io.BufferedWriter(f) # BufferedWriter creates a buffer between your Python code and the file.
    file.write(b"this is my buffer write\n")
    file.write(b"this is my second line that i am trying to write")
    file.flush()  # "Send the buffered data to the underlying file now."
# "test.text4" → file name
# "w" → write mode
# "b" → binary mode
# f → file object
```

```python
with open("test.text4","rb") as f:
    file = io.BufferedReader(f)
    data = file.read()
    print(data) 
```
