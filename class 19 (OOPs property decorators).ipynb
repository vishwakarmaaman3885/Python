#### Property Decorators - Getters, Setters, And Deletes

```python
# In Python, the @property decorator lets you access a method like an attribute.
# It is mainly used for encapsulation—controlling how an object's data is read, changed, or deleted.
# Getter = Get
# Setter = Set
# Deleter = Delete
```

```python
class course:
    def __init__(self,course_price,course_name):
        self.__course_price = course_price
        self.course_name = course_name
    
    @property
    def course_price_access(self):
        return self.__course_price
    
    @course_price_access.setter
    def course_price_set(self,price):
        if price<=3500:
            pass
        else:
            self.__course_price = price 
    
    @course_price_access.deleter
    def course_price_del(self):
        del self.__course_price
```

```python
price = course(5000,'Data_Science') 
```

```python
price.course_price_access
```

```python
price.course_name
```

```python
price.course_price_set=4500
```

```python
price.course_price_access
```

```python
del price.course_price_del
```
