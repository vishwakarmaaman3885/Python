```python
from functools import reduce # reduce() is used to reduce a sequence of elements to a single value by repeatedly applying a function.
```

```python
l = [1,2,3,4,5,4] 
```

```python
reduce(lambda x,y : x+y, l) 
```

```python
def add(x,y):
    return x+y 
```

```python
reduce(add,l) 
```

```python
reduce(lambda x,y,z : x+y+z, l)  
```
