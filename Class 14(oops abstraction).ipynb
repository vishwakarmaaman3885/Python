##### Abstraction - Abstraction means hiding internal details and exposing only the necessary functionality.

```python
# You know how to:
# Withdraw money
# Check balance
# But you don't know the internal banking process.

# This is abstraction

# Encapsulation → Hide data
# Abstraction → Hide implementation
# Inheritance → Reuse code
# Polymorphism → One method, many forms
```

```python
import abc

class college: # it's just creating a blueprint and call it where you need
    @abc.abstractmethod
    def student_details(self):
        pass
    @abc.abstractmethod
    def student_assignment(self):
        pass
    @abc.abstractmethod
    def student_marks(self):
        pass 
```

```python
class data_science(college):
    def student_details(self):
        return "it will try to return a details of data science masters"
    def student_assignment(self):
        return "it will return a details of student assignment for data science masters" 
```

```python
class web_dev(college):
   def student_details(self):
        return "this will return a details of web dev" 
   def student_marks(self):
       return "this will return a marks of web dev class" 
```

```python
ds = data_science()
ds.student_details()
```

```python
wb = web_dev()
wb.student_details() 
```
