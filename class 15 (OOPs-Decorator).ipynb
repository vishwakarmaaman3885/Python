#### Decorator(OOPs) - A decorator is a function that adds extra functionality to another function without changing its original code.

```python
def test():
    print("this is the start of my func")
    print(4+5)
    print("this is the end of my func")
```

```python
test() 
```

```python
def deco(func):
    def inner_deco(): # the inner function is usually used to wrap and modify the behavior of the original function.
        print("this is the start of my func")
        func()
        print('this is the end of my func')
    return inner_deco 
```

```python
@deco
def test1():
    print(7+8) 
```

```python
test1() 
```

```python
@deco
def test2():
    print(78+9)
```

```python
test2() 
```

```python
import time  # This is a timer decorator. Its purpose is to measure how much time a function takes to execute.
def timer_test(func):
    def timer_test_inner():
        start = time.time()
        func()
        end = time.time()
        print(end-start)
    return timer_test_inner
```

```python
@timer_test
def test3():
    print(67+89) 
```

```python
test3() 
```

```python
@timer_test
def test4():
    for i in range(10000):
        pass
```

```python
test4() 
```
