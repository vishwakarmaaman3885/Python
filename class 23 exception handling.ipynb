#### Exception handling is a way to handle errors during program execution so that your program doesn't stop unexpectedly.
For example, if you divide a number by zero, Python raises a ZeroDivisionError.

```python
# Exception                 Meaning

# ZeroDivisionError    Dividing by zero
# ValueError           Correct type of input expected, but invalid value supplied
# TypeError            Operation used with an incompatible type
# IndexError           List index is out of range
# KeyError             Dictionary key does not exist
# FileNotFoundError    File cannot be found
```

```python
# 1. try — code that may cause an error
# Python executes the code inside the try block first.

# 2. except — handles the error
# If an error occurs in try, Python executes the matching except block instead of stopping immediately.

#3. else-Executes if no error occurs in try.


# 4. finally — always executes
# The finally block runs whether an exception occurs or not.
```

#### Custom exception handling-Custom exception handling means creating your own error class to handle specific situations in your program.

```python
# For example, if you are creating a bank account program, you might want to raise a custom error when someone
# tries to withdraw more money than their balance.
```

```python
age = int(input("enter your age")) 
age
```

```python
class validateage(Exception):
    def __init__(self,msg):
        self.msg = msg 
```

```python
def validate_age(age):
    if age < 0: # raise is used to manually trigger an exception (error) when a particular condition occurs.
        raise validateage("age should not be lesser than zero")
    elif age > 200:
        raise validateage("age is too high")
    
    else:
        print("age is valid") 
```

```python
try : 
    age = int(input("enter your age"))
    validate_age(age)
except validateage as e : 
    print(e)
```

#### Best practice class

```python
# Use always a specific exception
try:
    10/0
except Exception as e:
    print(e) 
```

```python
try :
    10/0
except ZeroDivisionError as e:
    print(e) 
```

```python
# print always a valid message
try:
    10/0
except ZeroDivisionError as e:
    print("this is my zero divison error i am holding",e) 
```

```python
# always try to log
import logging
logging.basicConfig(filename="error.log",level=logging.ERROR)
try:
    10/0
except ZeroDivisionError as e:
    logging.error("this is my zero devision error i am handling {}".format(e)) 
```

```python
# always avoid to write a multiple exception handling
try:
    10/0
except FileNotFoundError as e:
    logging.error("this is my file not found {}".format(e))
except AttributeError as e:
    logging.error("this is my attribute error {}".format(e))
except ZeroDivisionError as e:
    logging.error("this is my zero devision error i am handling {}".format(e)) 
```
