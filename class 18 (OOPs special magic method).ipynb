#### Special Magic methods or dunder methods are special methods in Python whose names start and end with double underscores (__).

```python
# Dunder method	    Purpose
# __init__()	  Initialize object
# __str__()	      String representation
# __repr__()	  Developer representation
# __len__()	      len(object)
# __add__()	      object1 + object2
# __sub__()	      object1 - object2
# __eq__()	      object1 == object2
# __lt__()	      object1 < object2
# __gt__()	      object1 > object2
```
